Génération des données SOTRA

```python
import pandas as pd, numpy as np, random
random.seed(13); np.random.seed(13)

n = 70000
lignes = ['L1 Adjame-Treichville','L2 Yopougon-Plateau','L3 Cocody-Adjame',
          'L4 Marcory-Plateau','L5 Abobo-Adjame','L6 Port-Bouet-Plateau']
arrets = {l: [f'{l[:2]}-A{i:02d}' for i in range(1,random.randint(8,20))] for l in lignes}
bus_ids= [f'BUS{i:04d}' for i in range(1,151)]

rows = []
for _ in range(n):
    ligne = random.choice(lignes)
    heure = int(np.random.choice(range(5,23), p=np.array([2,3,4,6,4,3,3,4,5,6,5,4,3,3,4,6,5,3])/sum([2,3,4,6,4,3,3,4,5,6,5,4,3,3,4,6,5,3])))
    rush = heure in [7,8,17,18,19]
    rows.append({
        'id_voyage'     : f'VOY{_:07d}',
        'date_heure'    : pd.Timestamp('2024-01-01') + pd.Timedelta(days=random.randint(0,180), hours=heure, minutes=random.randint(0,59)),
        'ligne'         : ligne,
        'arret_depart'  : random.choice(arrets[ligne]),
        'bus_id'        : random.choice(bus_ids),
        'nb_passagers'  : int(np.clip(np.random.poisson(60 if rush else 30),0,120)),
        'capacite_bus'  : 80,
        'retard_min'    : int(np.clip(np.random.exponential(5 if rush else 2),0,45)),
        'incident'      : random.random() < 0.05,
        'type_incident' : random.choice(['Panne','Accident','Greve','']) if random.random()<0.05 else '',
    })

df = pd.DataFrame(rows)
df.loc[df.sample(frac=0.03).index,'nb_passagers'] = np.nan
df.to_csv('voyages_sotra_70k.csv', index=False)
print(f'Dataset SOTRA : {len(df):,} voyages')
```
